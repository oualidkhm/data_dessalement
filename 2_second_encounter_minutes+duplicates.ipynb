```python
import pandas as pd

# Lire le CSV
df_treated = pd.read_csv('df_treated.csv')

# S'assurer que 'time' est bien datetime
df_treated['time'] = pd.to_datetime(df_treated['time'])

# Mettre 'time' comme index
df_treated = df_treated.set_index('time')
```

```python
print(df_treated.index)
df_treated.head()
```

```python
# 1️⃣ Détecter les doublons exacts sur l'index
duplicate_time_mask = df_treated.index.duplicated(keep=False)
all_duplicated_rows = df_treated[duplicate_time_mask].sort_index()

print(f"Nombre de lignes exactes dupliquées : {len(all_duplicated_rows)}")
print(all_duplicated_rows.head(10))

# 2️⃣ Détecter les doublons par minute
time_floored_to_minute = df_treated.index.floor('T')
duplicate_minute_mask = time_floored_to_minute.duplicated(keep=False)
all_rows_same_minute = df_treated[duplicate_minute_mask].sort_index()

print(f"Nombre de lignes partageant la même minute : {len(all_rows_same_minute)}")
all_rows_same_minute.head(10)
```

```python
import pandas as pd
import numpy as np

print("Starting analysis of same-minute differences...")

# --- Step 0: travailler sur une copie ---
df = df_treated.copy()

# --- Step 1: créer la clé 'time_floored' depuis l'index ---
df['time_floored'] = df.index.floor('T')

# --- Step 2: trouver toutes les minutes ayant des doublons ---
minute_counts = df['time_floored'].value_counts()
duplicate_minutes = minute_counts[minute_counts > 1].index

# --- Step 3: isoler seulement les lignes de ces minutes ---
df_to_check = df[df['time_floored'].isin(duplicate_minutes)]

if df_to_check.empty:
    print("No rows found that share the same minute. Exiting.")
else:
    # --- Step 4: grouper et calculer min, max, mean ---
    grouped = df_to_check.groupby('time_floored')
    # colonnes numériques (capteurs) uniquement
    sensor_cols = df_to_check.select_dtypes(include=np.number).columns

    group_min = grouped[sensor_cols].min()
    group_max = grouped[sensor_cols].max()
    group_mean = grouped[sensor_cols].mean()

    # --- Step 5: différence absolue ---
    group_abs_diff = group_max - group_min

    # --- Step 6a: Détection de violations selon le prof (10% de la moyenne) ---
    group_threshold_mean10 = (group_mean + 1e-9).abs() * 0.10  # 10% de la mean (évite div0)
    is_problematic_mean10 = group_abs_diff > group_threshold_mean10

    # Compter violations par capteur (comme le prof)
    violation_counts = is_problematic_mean10.sum(axis=0).sort_values(ascending=False)
    final_counts = violation_counts[violation_counts > 0]
    print("\nColumns with the highest number of >10% differences (count):")
    if final_counts.empty:
        print("No columns found with >10% differences.")
    else:
        print(final_counts)

    # --- Step 6b: Calculer % diff relative au MIN (sécurisé) et afficher max% par capteur ---
    # éviter division par zéro / quasi-zero : clip les mins
    epsilon = 1e-6
    safe_min = group_min.clip(lower=epsilon)

    pct_diff_rel_min = (group_abs_diff / safe_min.abs()) * 100  # % relative to min
    # on prend le max % observed par capteur (pour détecter extrêmes)
    max_pct_diff_by_sensor = pct_diff_rel_min.max(axis=0).sort_values(ascending=False)
    max_pct_diff_by_sensor = max_pct_diff_by_sensor[max_pct_diff_by_sensor > 0]

    print("\nMax % Difference for sensors in 'same-minute' groups (relative to min, clipped):")
    print(max_pct_diff_by_sensor.head(30))  # afficher top 30

    # --- Step 6c: (optionnel) moyenne % diff relative au min par capteur ---
    mean_pct_diff_by_sensor = pct_diff_rel_min.mean(axis=0).sort_values(ascending=False)
    mean_pct_diff_by_sensor = mean_pct_diff_by_sensor[mean_pct_diff_by_sensor > 0]
    print("\nMean % Difference (relative to min) for sensors (top 20):")
    print(mean_pct_diff_by_sensor.head(20))

    # --- Step 7: affichage compact des minutes problématiques (min/max/mean) ---
    # définir minutes problématiques selon is_problematic_mean10 (prof style)
    problem_minutes = is_problematic_mean10.any(axis=1)[lambda s: s].index

    if len(problem_minutes) > 0:
        problematic_sensors = is_problematic_mean10.loc[problem_minutes].any(axis=0)
        problematic_cols = problematic_sensors[problematic_sensors].index.tolist()

        summary_df = pd.concat([
            group_min[problematic_cols].add_suffix('_min'),
            group_max[problematic_cols].add_suffix('_max'),
            group_mean[problematic_cols].add_suffix('_mean')
        ], axis=1)

        summary_df = summary_df.loc[problem_minutes].sort_index()
        print(f"\nFound {summary_df.shape[0]} minutes with >10% sensor differences (compact view):")
        print(summary_df.head(20))
    else:
        print("\nNo minutes found with a sensor difference > 10% (compact view).")

# --- Step 8: Aggregation by minute (mean of numeric sensors) ---
print("\nAggregating rows by minute (mean of numeric sensors)...")
df_minute_aggregated = df.groupby(df['time_floored']).mean(numeric_only=True).reset_index().rename(columns={'time_floored':'time'})
print("Aggregation complete. Preview:")
print(df_minute_aggregated.head())
```
