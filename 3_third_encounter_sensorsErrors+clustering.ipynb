Partie	Signification	Interprétation

➡️ La distribution est bimodale :
Une “mode” à 0 (arrêt / tests / pannes)
Une autre vers 200 (production effective)

### 🧠 Détection automatique des différents cas (arrêts, tests, pannes...)

On peut appliquer un traitement de **clustering** pour distinguer les différents comportements du système à partir des données mesurées.

#### 💡 Pourquoi sur les mesures ?

Deux approches possibles :

1️⃣ **Analyse simple (rapide)**  
   - Tu appliques le clustering **seulement sur la variable du débit**.  
   - Cela permet de repérer :
     - Le **cluster autour de 0** → arrêts, tests, ou capteur inactif.  
     - Le **cluster autour de 200** → fonctionnement normal.  
   👉 Idéal pour une première analyse rapide.

2️⃣ **Analyse avancée (multi-capteurs)**  
   - Tu appliques le clustering sur **plusieurs mesures** (ex. débit, pression, pH, conductivité...).  
   - Cela permet de mieux distinguer :
     - Un **arrêt normal ou planifié**,  
     - Une **panne ou un dysfonctionnement**,  
     - Une **phase de test manuelle**.

#### ⚙️ En résumé

| Objectif | Type d'analyse | Variables utilisées | Résultat |
|-----------|----------------|----------------------|-----------|
| Rapide | Simple | 1 variable (débit) | Détection ON/OFF |
| Précise | Avancée | Plusieurs capteurs | Détection de pannes, tests, arrêts |

```python
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# 1️⃣ Copier le DataFrame pour test
df_test = df_minute_aggregated.copy()

# 2️⃣ Sélection de la variable
flow_col = 'FIT_005F_PERMEATE FLOW'
X = df_test[[flow_col]].dropna()

# 3️⃣ Clustering (2 groupes : arrêt / marche)
kmeans = KMeans(n_clusters=2, random_state=42)
X['cluster'] = kmeans.fit_predict(X)

# 4️⃣ Centres des clusters
centers = kmeans.cluster_centers_
print("Centres des clusters :", centers.flatten())

# 5️⃣ Visualisation
plt.figure(figsize=(12,4))
plt.scatter(df_test.index, df_test[flow_col], c=X['cluster'], cmap='coolwarm', s=5)
plt.title("Clustering du débit perméat (FIT_005F_PERMEATE FLOW)")
plt.xlabel("Temps")
plt.ylabel("Débit (flow)")
plt.show()

```

```python
# pour comprendre, on cherche les zeros qui ne sont pas vrais non pas seulement en cette colonne (le deuxième mode en majorité des histogrammes),
# erreur de capteur ou test. Ces zéros sont détectables car, dans une ligne donnée, une colonne contient un zéro
# tandis que les autres ont des valeurs éloignées de zéro. Comme la majorité des graphes sont asymétriques vers
# la droite, on choisit de remplacer ces zéros suspects par la médiane de leur colonne (valeur la plus crédible
# quand la loi normale est décalée vers la droite).

df_corrected = df_minute_aggregated.copy()

# Seuils
zero_threshold = 0.05
ratio_threshold = 0.2

# Colonnes numériques
num_cols = df_corrected.select_dtypes(include=[np.number]).columns
medians = df_corrected[num_cols].median()

# Matrice booléenne : chaque valeur proche de zéro ?
near_zero = abs(df_corrected[num_cols] - 0) < (medians * zero_threshold)

# Matrice booléenne : chaque valeur "grande" ?
large_values = df_corrected[num_cols] > (medians * ratio_threshold)

# Pour chaque ligne et colonne, vérifier si la majorité des autres colonnes sont grandes
mask_replace = pd.DataFrame(False, index=df_corrected.index, columns=num_cols)

for col in num_cols:
    # Ratio de grandes valeurs dans les autres colonnes
    other_cols = [c for c in num_cols if c != col]
    ratio = large_values[other_cols].sum(axis=1) / len(other_cols)
    mask_replace[col] = near_zero[col] & (ratio > 0.8)

# Remplacer par la médiane
for col in num_cols:
    df_corrected.loc[mask_replace[col], col] = medians[col]

# Vérif
before = (abs(df_minute_aggregated[num_cols]) < medians * zero_threshold).sum().sum()
after = (abs(df_corrected[num_cols]) < medians * zero_threshold).sum().sum()
print(f"Valeurs proches de zéro avant correction : {before}")
print(f"Valeurs proches de zéro après correction  : {after}")
print(f"→ {before - after} valeurs suspectes remplacées par la médiane correspondante.")
```

```python
import numpy as np
import pandas as pd

# === Seuils définis ===
seuils = {
    'hmmf_inlet_flow' : (200, 650, 0),
    'hmmf_diff_pressure' : (0, 1.5, 0),
    'ro_discharge_pressure_hp_pmp': (10, 70, 0),
    'ro_reject_line_pressure': (10, 65, 0),
    'ro_permeate_ph': (5, 8, 7),
    'ro_permeate_cond': (150, 1300, 400),
    'ro_permeate_flow': (50, 250, 0),
    'ro_cartridge_outflow': (50, 250, 0),
    'ro_inlet_px_flow': (100, 500, 0),
    'ph_feed_hmmf':(6, 8.5, 7),
    'pit_feed_hmmf':(1, 3.5, 0),
    'ph_feed_ro':(5.5, 8.5, 7),
    'pit_feed_ro':(3, 5.5, 0),
    'orp_feed_ro':(100, 650, 400),
    'cond_feed_ro':(25, 60, 45),
    'INTAKE_LINE_PRESSURE_OF_HMMF_FEED_PUMPS' : (0.5, 2.5, 0),
    'BACKWASH_PUMP_FLOW' : (0, 500, 0),
    'LIT_003_OF_HMMF_BACKWASH_STORAGE_TANK' : (15, 40,0),
    'LIT_002_FILTER_WATER_STG_TNK': (0, 100,0),
}

# === Fonction pour détecter la famille selon le nom de la colonne ===
def detect_family(col_name):
    name = col_name.upper()
    if "FLOW AT INLET HMMF" in name:
        return 'hmmf_inlet_flow'
    elif "DIFF PRESSURE ACROSS HMMF" in name:
        return 'hmmf_diff_pressure'
    elif "DISCHARGE_PRESSURE_HP PMP" in name:
        return 'ro_discharge_pressure_hp_pmp'
    elif "REJECT LINE PRESSURE" in name:
        return 'ro_reject_line_pressure'
    elif "_PH" in name:
        if "RO FEEDPUMP" in name:
            return 'ph_feed_ro'
        elif "DISCHARGE HMMF FEED PUMPS" in name:
            return 'ph_feed_hmmf'
        else:
            return 'ro_permeate_ph'
    elif "_COND" in name:
        if "RO FEEDPUMP" in name:
            return 'cond_feed_ro'
        else:
            return 'ro_permeate_cond'
    elif "PERMEATE FLOW" in name:
        return 'ro_permeate_flow'
    elif "CARTRIDGE_OUTFLOW" in name:
        return 'ro_cartridge_outflow'
    elif "INLET PX_FLOW" in name:
        return 'ro_inlet_px_flow'
    elif "ORP AT RO FEEDPUMP" in name:
        return 'orp_feed_ro'
    elif "INTAKE LINE PRESSURE OF HMMF FEED PUMPS" in name:
        return 'INTAKE_LINE_PRESSURE_OF_HMMF_FEED_PUMPS'
    elif "BACKWASH PUMP FLOW" in name:
        return 'BACKWASH_PUMP_FLOW'
    elif "OF HMMF BACKWASH STORAGE TANK" in name:
        return 'LIT_003_OF_HMMF_BACKWASH_STORAGE_TANK'
    elif "FILTER_WATER_STG_TNK" in name:
        return 'LIT_002_FILTER_WATER_STG_TNK'
    elif "DISCHARGE PRESSURE AT HMMF" in name:
        return 'pit_feed_hmmf'
    elif "RO FEEDPUMP_P2" in name:
        return 'pit_feed_ro'
    else:
        return None

# === Appliquer les seuils ===
unknown_cols = []

for col in df_corrected.columns:
    family = detect_family(col)
    if family and family in seuils:
        min_val, max_val, default = seuils[family]
        # Remplacer valeurs hors limites par default
        df_corrected[col] = np.where((df_corrected[col] < min_val) | (df_corrected[col] > max_val),
                                     default, df_corrected[col])
    else:
        unknown_cols.append(col)

# === Afficher les colonnes non reconnues ===
if unknown_cols:
    print("Colonnes sans famille détectée :")
    for col in unknown_cols:
        print(col)
else:
    print("Toutes les colonnes ont été traitées.")
```

```python
from sklearn.cluster import KMeans
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

original_columns = df_corrected.columns.tolist()

# Sélection aléatoire de 5 colonnes
sample_cols = np.random.choice([col for col in original_columns if not col.endswith('_cluster')], size=5, replace=False)

# === Création des clusters pour toutes les colonnes (temporairement) ===
for col in original_columns:
    if col.endswith('_cluster'):
        continue

    X = df_corrected[[col]].values
    kmeans = KMeans(n_clusters=3, random_state=42, n_init=5)
    clusters = kmeans.fit_predict(X)
    cluster_col = f'{col}_cluster'
    df_corrected[cluster_col] = clusters

# === Affichage avant correction avec couleurs par cluster ===
for col in sample_cols:
    cluster_col = f'{col}_cluster'
    plt.figure(figsize=(8,4))
    sns.histplot(data=df_corrected, x=col, hue=cluster_col, palette='Set2', bins=50, kde=True)
    plt.title(f'Distribution avant correction de {col}')
    plt.show()

# === Correction des clusters contenant 0 ===
for col in original_columns:
    if col.endswith('_cluster'):
        continue
    cluster_col = f'{col}_cluster'
    cluster_with_zero = df_corrected.loc[df_corrected[col] == 0, cluster_col].unique()
    df_corrected.loc[df_corrected[cluster_col].isin(cluster_with_zero), col] = 0

# === Affichage après correction avec couleurs par cluster ===
for col in sample_cols:
    cluster_col = f'{col}_cluster'
    plt.figure(figsize=(8,4))
    sns.histplot(data=df_corrected, x=col, hue=cluster_col, palette='Set2', bins=50, kde=True)
    plt.title(f'Distribution après correction de {col}')
    plt.show()

# === Supprimer les colonnes cluster pour revenir à 91 colonnes ===
df_corrected = df_corrected[[col for col in df_corrected.columns if not col.endswith('_cluster')]]
```
