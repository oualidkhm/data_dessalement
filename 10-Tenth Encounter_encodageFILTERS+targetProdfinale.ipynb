```python
import pandas as pd
import ast  # pour convertir les chaînes de liste en liste Python si nécessaire

# 🔹 Si les valeurs sont des chaînes comme "['HMMF-A', 'HMMF-C']", les convertir en liste
df['FILTRES_A_BACKWASHER'] = df['FILTRES_A_BACKWASHER'].apply(ast.literal_eval)

# 🔹 Trouver tous les filtres uniques
all_filters = sorted({f for sublist in df['FILTRES_A_BACKWASHER'] for f in sublist})

# 🔹 Créer les colonnes binaires
for f in all_filters:
    df[f] = df['FILTRES_A_BACKWASHER'].apply(lambda x: 1 if f in x else 0)

# 🔹 Vérifier
print(df[['FILTRES_A_BACKWASHER'] + all_filters].head())
```

```python

# 🔹 Supprimer la colonne d'origine si tu veux
df = df.drop(columns=['FILTRES_A_BACKWASHER'])
```

```python
# 🔹 Renommer Production_totale en target
df = df.rename(columns={'Production_totale': 'target'})

# 🔹 Déplacer target à la fin
cols = [c for c in df.columns if c != 'target'] + ['target']
df = df[cols]

# 🔹 Vérifier
print(df.columns)
```
